## 1. Introduction

Buildings play an important role in urban energy consumption and climate policy. Understanding how much energy buildings consume relative to their size can help identify patterns in energy use and highlight potential data quality concerns.

This project explores the distribution of **Site Energy Use Intensity (Site EUI)** among buildings reporting energy benchmarking data in New York City.

Site EUI measures annual energy consumption per square foot of building area, expressed in kBtu/ft². Lower values generally indicate less energy consumed per unit of floor area, although comparisons should account for differences in building type and use.

The analysis addresses three questions:

1. What are the mean, median, and mode of reported Site EUI?
2. How do unusually high values influence the interpretation of these statistics?
3. How are building records distributed across different EUI ranges?

Using both Pandas and the Python Standard Library, the project demonstrates two approaches to descriptive statistical analysis and presents the findings through a simple visualization.

## 6. Recalculating Statistics Using the Python Standard Library

To verify the results obtained using Pandas, I independently calculated the mean, median, and mode using only Python's standard library.

Unlike the previous approach, this section reads the original CSV file directly using the built-in `csv` module. It does not access the Pandas DataFrame or rely on Pandas' statistical functions.

The same data-cleaning criteria were applied: only positive, numeric Site EUI values were retained.



```python
import csv
from pathlib import Path

eui_values = []

data_path = next(Path(".").glob("NYC_Building_Energy*.csv"))

with open(data_path, newline="", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)

    for row in reader:
        value = row["Site EUI (kBtu/ft²)"]

        try:
            number = float(value)
            if number > 0:
                eui_values.append(number)
        except ValueError:
            continue

print(len(eui_values))
```

### 6.1 Calculating the Mean

The mean was calculated by summing all valid EUI values and dividing by the number of observations.



```python
total = 0

for value in eui_values:
    total += value

mean_manual = total / len(eui_values)

print("Manual Mean EUI:", mean_manual)
```

### 6.2 Calculating the Median

The observations were sorted in ascending order. For an odd number of observations, the middle value was selected. For an even number, the average of the two middle values was calculated.


```python
sorted_eui = sorted(eui_values)
n = len(sorted_eui)

if n % 2 == 1:
    median_manual = sorted_eui[n // 2]
else:
    middle1 = sorted_eui[n // 2 - 1]
    middle2 = sorted_eui[n // 2]
    median_manual = (middle1 + middle2) / 2

print("Manual Median EUI:", median_manual)
```

### 6.3 Calculating the Mode

A dictionary was used to count how frequently each EUI value appeared. The value or values with the highest frequency were identified as the mode.


```python
counts = {}

for value in eui_values:
    if value in counts:
        counts[value] += 1
    else:
        counts[value] = 1

max_count = max(counts.values())

mode_manual = []

for value, count in counts.items():
    if count == max_count:
        mode_manual.append(value)

mode_manual.sort()

print("Manual Mode EUI:", mode_manual)
print("Frequency:", max_count)
```

### Comparison with Pandas

Both approaches produced consistent results:

| Statistic | Pandas | Standard Library |
|---|---:|---:|
| Mean | 7,029.21 | 7,029.21 |
| Median | 71.5 | 71.5 |
| Mode | 68.6, 75.6 | 68.6, 75.6 |

The matching results confirm that the manual calculations are consistent with the Pandas implementation. Minor differences in the unrounded mean are attributable to floating-point precision.

## 7. Visualizing the Distribution of Building Energy Use

To better understand the distribution of Site EUI, I created a horizontal bar chart using only the Python Standard Library.

The observations were grouped into six EUI ranges, from 0–50 to over 1,000 kBtu/ft². Each bar represents the number of building-year records within a particular range.

Rather than using an external plotting library, the visualization uses Python's string repetition to generate bars whose lengths are proportional to the number of observations.

### Interpreting the Distribution

The chart reveals that most observations fall within relatively low EUI ranges, with the largest concentration between 50 and 100 kBtu/ft².

In contrast, relatively few observations exceed 1,000 kBtu/ft². However, some of these records contain extremely large values, which substantially influence the arithmetic mean.

This contrast helps explain why the median provides a more representative measure of a typical building-year observation than the mean in this dataset.

```python
# Group EUI values into ranges
ranges = {
    "0–50": 0,
    "50–100": 0,
    "100–200": 0,
    "200–500": 0,
    "500–1000": 0,
    "Over 1000": 0,
}

for value in eui_values:
    if value < 50:
        ranges["0–50"] += 1
    elif value < 100:
        ranges["50–100"] += 1
    elif value < 200:
        ranges["100–200"] += 1
    elif value < 500:
        ranges["200–500"] += 1
    elif value <= 1000:
        ranges["500–1000"] += 1
    else:
        ranges["Over 1000"] += 1

# Draw the chart
max_count = max(ranges.values())

print("NYC Building Energy Use Intensity Distribution")
print("EUI (kBtu/ft²) | Number of building-year records\n")

for category, count in ranges.items():
    bar_length = round(count / max_count * 40)
    bar = "✨" * bar_length
    print(f"{category:>10} | {bar} {count:,}")
```

## 8. Key Findings and Limitations

### 8.1 Key Findings

**Finding 1: Most building-year observations are concentrated within a relatively narrow EUI range.**

Among the 90,078 valid observations, 50,898 (approximately 56.5%) fall between 50 and 100 kBtu/ft². The median Site EUI is 71.5 kBtu/ft², suggesting that typical reported energy use is substantially lower than the arithmetic mean.

**Finding 2: Extreme values substantially influence the mean.**

The mean Site EUI is approximately 7,029.21 kBtu/ft², nearly 98 times the median. Several observations report EUI values exceeding 10 million kBtu/ft².

These extreme observations demonstrate how a small number of unusually large values can distort summary statistics. For this dataset, the median offers a more informative description of the typical observation.

**Finding 3: Data cleaning is essential for reliable analysis.**

Of the 103,259 original records, 12,904 could not be converted into numeric EUI values. An additional 277 records contained zero or negative values.

After applying consistent cleaning criteria, 90,078 observations remained. Both Pandas and standard-library calculations produced matching descriptive statistics.

### 8.2 Limitations

Several limitations should be considered when interpreting the findings.

1. **Reporting accuracy:** The dataset contains extremely high EUI values that may reflect reporting, measurement, or calculation issues. Their underlying causes have not been independently verified.

2. **Building heterogeneity:** Different building types have different energy requirements. A hospital, office building, and residential property should not necessarily be evaluated using identical benchmarks.

3. **Multiple reporting years:** The dataset includes observations from multiple years. Individual properties may appear more than once, meaning the results describe building-year records rather than unique buildings.

4. **Incomplete coverage:** The dataset primarily covers properties subject to NYC's energy benchmarking requirements. Therefore, the findings should not be generalized to every building in New York City.

5. **Limited analytical scope:** This project focuses on descriptive statistics and does not establish causal relationships between building characteristics, energy consumption, or policy outcomes.

### 8.3 Conclusion

This analysis illustrates the importance of combining basic statistical methods with careful data interpretation.

Although the mean Site EUI is approximately 7,029.21 kBtu/ft², the median is only 71.5 kBtu/ft², highlighting the substantial influence of extreme observations.

From a policy perspective, these findings suggest that reliable energy benchmarking depends not only on collecting building-level data but also on identifying potential data quality issues and selecting appropriate summary measures.
